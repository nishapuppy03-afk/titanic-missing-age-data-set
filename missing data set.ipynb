```python
import pandas as pd

df = pd.read_csv("test[1].csv")
print(df.isnull())
```

```python

print(df.isnull().sum())
```

```python

df_dropna = df.dropna()
print(df_dropna.head())
```

```python

df_fillna = df.fillna(0)
print(df_fillna.head())
```

```python

df['Age_mean_filled'] = df['Age'].fillna(df['Age'].mean())
print(df[['Age', 'Age_mean_filled']].head())
```

```python

df['Age_median_filled'] = df['Age'].fillna(df['Age'].median())
print(df[['Age', 'Age_median_filled']].head())
```

```python

df['Sex_encoded'] = df['Sex'].map({'male': 0, 'female': 1})
print(df[['Sex', 'Sex_encoded']].head())
```
